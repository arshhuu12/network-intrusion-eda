```python
import sys
import pandas as pd
import numpy as np
import matplotlib
import matplotlib.pyplot as plt
import seaborn as sns
import scipy
import sklearn

print("Python executable:", sys.executable)
print("Python version:", sys.version.split()[0])

print("\nLibrary versions:")
print("Pandas:", pd.__version__)
print("NumPy:", np.__version__)
print("Matplotlib:", matplotlib.__version__)
print("Seaborn:", sns.__version__)
print("SciPy:", scipy.__version__)
print("Scikit-learn:", sklearn.__version__)

assert ".venv" in sys.executable.lower(), "Wrong Python environment!"

print("\nSUCCESS: Project environment verified!")
```

```python
# Small sample dataset for environment testing

sample_data = pd.DataFrame({
    "traffic_type": ["Normal", "Attack", "Normal", "Attack", "Normal"],
    "duration": [10, 45, 15, 90, 20],
    "bytes": [100, 500, 150, 900, 200]
})

print("Dataset shape:", sample_data.shape)

display(sample_data)

print("\nSummary statistics:")
display(sample_data.describe())
```

```python
from pathlib import Path

# Create a sample visualization
plt.figure(figsize=(7, 4))

sns.countplot(
    data=sample_data,
    x="traffic_type"
)

plt.title("Sample Network Traffic Distribution")
plt.xlabel("Traffic Type")
plt.ylabel("Number of Records")

plt.tight_layout()

# Save the figure
output_path = Path("../figures/environment_test.png")
output_path.parent.mkdir(parents=True, exist_ok=True)

plt.savefig(output_path, dpi=150)

plt.show()

print("Visualization saved successfully!")
```
