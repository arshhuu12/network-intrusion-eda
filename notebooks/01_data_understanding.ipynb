```python
import pandas as pd
from pathlib import Path

print("Pandas version:", pd.__version__)
print("Current working directory:", Path.cwd())
```

```python
from pathlib import Path

current_dir = Path.cwd()

if current_dir.name == "notebooks":
    project_dir = current_dir.parent
else:
    project_dir = current_dir

print("Project directory:", project_dir)

nsl_train_path = project_dir / "data" / "raw" / "nsl_kdd" / "KDDTrain+.txt"
nsl_test_path = project_dir / "data" / "raw" / "nsl_kdd" / "KDDTest+.txt"

print("Training file exists:", nsl_train_path.exists())
print("Testing file exists:", nsl_test_path.exists())

assert nsl_train_path.is_file(), "NSL-KDD training file not found!"
assert nsl_test_path.is_file(), "NSL-KDD testing file not found!"
```

```python
nsl_columns = [
    "duration",
    "protocol_type",
    "service",
    "flag",
    "src_bytes",
    "dst_bytes",
    "land",
    "wrong_fragment",
    "urgent",
    "hot",
    "num_failed_logins",
    "logged_in",
    "num_compromised",
    "root_shell",
    "su_attempted",
    "num_root",
    "num_file_creations",
    "num_shells",
    "num_access_files",
    "num_outbound_cmds",
    "is_host_login",
    "is_guest_login",
    "count",
    "srv_count",
    "serror_rate",
    "srv_serror_rate",
    "rerror_rate",
    "srv_rerror_rate",
    "same_srv_rate",
    "diff_srv_rate",
    "srv_diff_host_rate",
    "dst_host_count",
    "dst_host_srv_count",
    "dst_host_same_srv_rate",
    "dst_host_diff_srv_rate",
    "dst_host_same_src_port_rate",
    "dst_host_srv_diff_host_rate",
    "dst_host_serror_rate",
    "dst_host_srv_serror_rate",
    "dst_host_rerror_rate",
    "dst_host_srv_rerror_rate",
    "attack_type",
    "difficulty_level"
]

print("Number of column names:", len(nsl_columns))

assert len(nsl_columns) == 43
```

```python
nsl_train = pd.read_csv(
    nsl_train_path,
    header=None,
    names=nsl_columns
)

nsl_test = pd.read_csv(
    nsl_test_path,
    header=None,
    names=nsl_columns
)

print("NSL-KDD Training:", nsl_train.shape)
print("NSL-KDD Testing:", nsl_test.shape)
```

```python
display(nsl_train.head())
```

```python
nsl_train[
    ["duration", "protocol_type", "service",
     "src_bytes", "dst_bytes", "attack_type", "difficulty_level"]
].head()
```

```python
print("NSL-KDD Training Data Types:")
print(nsl_train.dtypes.to_string())
```

```python
numeric_columns = nsl_train.select_dtypes(
    include="number"
).columns.tolist()

categorical_columns = nsl_train.select_dtypes(
    include=["object", "str", "category"]
).columns.tolist()

print("Numerical columns:", len(numeric_columns))
print("Categorical columns:", len(categorical_columns))

print("\nCategorical columns:")
print(categorical_columns)
```

```python
print("Number of unique attack labels:")
print(nsl_train["attack_type"].nunique())

print("\nAttack labels:")
print(sorted(nsl_train["attack_type"].unique()))
```

```python
nsl_train["traffic_class"] = nsl_train["attack_type"].apply(
    lambda x: "Normal" if x == "normal" else "Attack"
)

print("Training traffic distribution:")
print(nsl_train["traffic_class"].value_counts())

print("\nTraining traffic percentages:")
print(
    nsl_train["traffic_class"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)
```

```python
train_attacks = set(nsl_train["attack_type"].unique())
test_attacks = set(nsl_test["attack_type"].unique())

print("Unique labels in training:", len(train_attacks))
print("Unique labels in testing:", len(test_attacks))

print("\nLabels found only in testing:")
print(sorted(test_attacks - train_attacks))

print("\nLabels found only in training:")
print(sorted(train_attacks - test_attacks))
```

```python
## UNSW-NB15 Dataset Understanding
```

```python
unsw_train_path = project_dir / "data" / "raw" / "unsw_nb15" / "UNSW_NB15_training-set.csv"
unsw_test_path = project_dir / "data" / "raw" / "unsw_nb15" / "UNSW_NB15_testing-set.csv"

unsw_train = pd.read_csv(unsw_train_path)
unsw_test = pd.read_csv(unsw_test_path)

print("UNSW-NB15 Training:", unsw_train.shape)
print("UNSW-NB15 Testing:", unsw_test.shape)

assert unsw_train.shape == (175341, 45)
assert unsw_test.shape == (82332, 45)
```

```python
print("UNSW-NB15 Column Names:\n")

for i, column in enumerate(unsw_train.columns, start=1):
    print(f"{i:2}. {column}")
```

```python
print("UNSW-NB15 Data Types:\n")
print(unsw_train.dtypes.to_string())

print("\nNumerical columns:")
print(len(unsw_train.select_dtypes(include="number").columns))

print("\nCategorical columns:")
print(len(unsw_train.select_dtypes(include=["object", "str", "category"]).columns))
```

```python
print("UNSW-NB15 Training Class Distribution:\n")

print(unsw_train["label"].value_counts().sort_index())

print("\nClass Percentages:\n")

print(
    unsw_train["label"]
    .value_counts(normalize=True)
    .sort_index()
    .mul(100)
    .round(2)
)
```

```python
print("UNSW-NB15 Attack Categories:\n")

print(
    unsw_train["attack_cat"]
    .value_counts(dropna=False)
    .to_string()
)
```

```python
train_categories = set(unsw_train["attack_cat"].dropna().unique())
test_categories = set(unsw_test["attack_cat"].dropna().unique())

print("Training categories:", len(train_categories))
print("Testing categories:", len(test_categories))

print("\nCategories only in testing:")
print(sorted(test_categories - train_categories))

print("\nCategories only in training:")
print(sorted(train_categories - test_categories))
```

```python
dataset_summary = pd.DataFrame([
    {
        "Dataset": "NSL-KDD",
        "Training Rows": len(nsl_train),
        "Testing Rows": len(nsl_test),
        "Total Rows": len(nsl_train) + len(nsl_test),
        "Columns": len(nsl_train.columns) - 1,
        "Categorical Features": 3,
        "Numerical Features": 38,
        "Training Attack %": round(
            (nsl_train["attack_type"] != "normal").mean() * 100, 2
        )
    },
    {
        "Dataset": "UNSW-NB15",
        "Training Rows": len(unsw_train),
        "Testing Rows": len(unsw_test),
        "Total Rows": len(unsw_train) + len(unsw_test),
        "Columns": len(unsw_train.columns),
        "Categorical Features": 3,
        "Numerical Features": 39,
        "Training Attack %": round(
            (unsw_train["label"] == 1).mean() * 100, 2
        )
    }
])

display(dataset_summary)
```

```python
output_path = project_dir / "reports" / "dataset_summary.csv"

dataset_summary.to_csv(output_path, index=False)

print("Dataset summary saved to:", output_path)
```
