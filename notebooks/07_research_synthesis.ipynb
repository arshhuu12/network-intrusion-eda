## M8.2 — Research Insight Development

### Objective

Transform verified exploratory findings into meaningful
research insights.

Each insight contains:

1. Observation — What the data shows.
2. Interpretation — What the observation means.
3. Implication — Why the finding matters.
4. Limitation — What cannot be concluded.

The analysis focuses on dataset quality, attack diversity,
train-test differences, and feature relationships.

## M8.3 — Final Comparative Research Visualizations

### Objective

Create three presentation-quality figures summarizing:

1. Train–test attack proportions.
2. Categorical feature distribution differences.
3. Attack-label coverage across partitions.

The figures use results validated in earlier milestones.

No additional machine-learning analysis is performed.

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

final_figures_dir = (
    project_dir / "figures" / "comparative"
)

final_figures_dir.mkdir(
    parents=True,
    exist_ok=True
)

sns.set_theme(style="whitegrid")

print("Final figures directory:", final_figures_dir)
```

```python
attack_proportion_data = pd.DataFrame({
    "Dataset": [
        "NSL-KDD", "NSL-KDD",
        "UNSW-NB15", "UNSW-NB15"
    ],
    "Partition": [
        "Train", "Test",
        "Train", "Test"
    ],
    "Attack Percentage": [
        46.54, 56.92,
        68.06, 55.06
    ]
})

plt.figure(figsize=(10, 6))

ax = sns.barplot(
    data=attack_proportion_data,
    x="Dataset",
    y="Attack Percentage",
    hue="Partition",
    errorbar=None
)

for container in ax.containers:
    ax.bar_label(
        container,
        fmt="%.2f%%",
        padding=3
    )

plt.title(
    "Attack Prevalence: Training vs Testing",
    fontsize=15,
    fontweight="bold"
)

plt.ylabel("Attack Records (%)")
plt.xlabel("Dataset")
plt.ylim(0, 85)

plt.tight_layout()

figure1_path = (
    final_figures_dir
    / "final_attack_prevalence_comparison.png"
)

plt.savefig(
    figure1_path,
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("Figure 1 saved:", figure1_path.exists())
```

```python
tvd_data = pd.DataFrame({
    "Dataset": [
        "NSL-KDD",
        "NSL-KDD",
        "NSL-KDD",
        "UNSW-NB15",
        "UNSW-NB15",
        "UNSW-NB15"
    ],
    "Feature": [
        "service",
        "flag",
        "protocol_type",
        "proto",
        "state",
        "service"
    ],
    "TVD": [
        0.2049,
        0.1906,
        0.0223,
        0.0677,
        0.0548,
        0.0357
    ]
})

tvd_data["Feature Label"] = (
    tvd_data["Dataset"]
    + " — "
    + tvd_data["Feature"]
)

tvd_data = tvd_data.sort_values(
    "TVD",
    ascending=False
)

plt.figure(figsize=(12, 6))

ax = sns.barplot(
    data=tvd_data,
    x="TVD",
    y="Feature Label",
    hue="Dataset",
    dodge=False
)

for container in ax.containers:
    ax.bar_label(
        container,
        fmt="%.4f",
        padding=3
    )

plt.title(
    "Train–Test Categorical Distribution Differences",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("Total Variation Distance (TVD)")
plt.ylabel("Dataset and Feature")

plt.xlim(0, 0.25)

plt.tight_layout()

figure2_path = (
    final_figures_dir
    / "final_categorical_shift_comparison.png"
)

plt.savefig(
    figure2_path,
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("Figure 2 saved:", figure2_path.exists())
```

```python
attack_coverage_data = pd.DataFrame({
    "Dataset": [
        "NSL-KDD",
        "NSL-KDD",
        "NSL-KDD",
        "UNSW-NB15",
        "UNSW-NB15",
        "UNSW-NB15"
    ],
    "Coverage": [
        "Shared",
        "Test-only",
        "Train-only",
        "Shared",
        "Test-only",
        "Train-only"
    ],
    "Count": [
        20, 17, 2,
        9, 0, 0
    ]
})

plt.figure(figsize=(11, 6))

ax = sns.barplot(
    data=attack_coverage_data,
    x="Dataset",
    y="Count",
    hue="Coverage",
    errorbar=None
)

for container in ax.containers:
    ax.bar_label(
        container,
        fmt="%.0f",
        padding=3
    )

plt.title(
    "Attack-Label Coverage Across Train and Test",
    fontsize=15,
    fontweight="bold"
)

plt.ylabel("Number of Attack Labels")
plt.xlabel("Dataset")

plt.ylim(0, 25)

plt.tight_layout()

figure3_path = (
    final_figures_dir
    / "final_attack_label_coverage.png"
)

plt.savefig(
    figure3_path,
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("Figure 3 saved:", figure3_path.exists())
```

```python
final_figure_paths = [
    figure1_path,
    figure2_path,
    figure3_path
]

assert len(final_figure_paths) == 3

for path in final_figure_paths:

    assert path.exists(), (
        f"Missing figure: {path.name}"
    )

    assert path.stat().st_size > 0, (
        f"Empty figure: {path.name}"
    )

    print(
        f"{path.name}: "
        f"{path.stat().st_size:,} bytes"
    )

print(
    "\nSUCCESS: M8.3 final comparative "
    "visualizations validated!"
)
```
